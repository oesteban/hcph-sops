# Development of the Bayesian model to characterize the reliability of SC on synthetic data

Thinking about the problem, I converged to the following model of the density distribution:

p(D) = p(D| c=0)p(c=0) + p(D| c=1)p(c=1)

     = pi_0 * Exp(lambda) + (1-pi_0) * TruncatedNormal(mu, sigma)

with 

pi_0 ~ Beta(1,1)

lambda ~ Gamma(2,3)

sigma ~ HalfNormal(0,0.1)

```python
from bayesian_modeling import run_simulation
results_fixed = run_simulation(true_pi0=0.3, true_lambda=2.0, true_mu=0.8, true_sigma=0.15)
```

```python

# Run with mu as a learned parameter
results_learned = run_simulation(
     true_pi0=0.3, true_lambda=2.0, true_mu=0.8, 
     mu_type="learned", mu_prior_mean=0.8, mu_prior_sigma=0.2
)
```
